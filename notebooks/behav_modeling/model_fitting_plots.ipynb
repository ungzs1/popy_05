# Load behavior

```python
for monkey in ['ka', 'po', 'yu_sham', 'yu_DCZ']:
    behav = pd.read_pickle(f'results/simulation_behaviors_{monkey}.pkl')
    behav['monkey'] = [f'{monkey} {model}' for monkey, model in zip(behav['monkey'], behav['model'])]

    behav = behav.reset_index(drop=True)
    behav = behav.drop(columns=['model'])

    behav = convert_column_format(behav, original='simulation')

    behav = add_history_of_feedback(behav, num_trials=3, one_column=False, coding=(0, 1), binary=False)
    behav = add_switch_info(behav, add_trials_since_switch=True, flip_coding=False)
    #behav = _add_trial_since_switch(behav)
    behav = behav.dropna()

    #results = pd.read_csv(f'results/simulation_results_{monkey}.csv')

    #behav_temp = behavs.loc[behavs.f'MONKEY {monkey.upper()}', 'Repeating agent', 'WSLS agent (long history)']}
    plot_hist_thingy(behav, title='Baseline models VS Monkey {}'.format(monkey.upper()))
    plot_strategy(behav, title='Baseline models VS Monkey {}'.format(monkey.upper()))
```
