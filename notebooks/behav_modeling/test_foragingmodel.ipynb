```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
import pandas as pd

from popy.io_tools import load_behavior
from popy.behavior_data_tools import (
    drop_time_fields,
    add_foraging_value,
    add_switch_info,
)

# load monkey behavior data
behav = load_behavior()
behav = drop_time_fields(behav)  # remove unnecessary attributes
behav = behav.dropna(subset=['target'])

# compute foraging value for each trial and add it to the dataframe
behav = add_foraging_value(behav)
# get is monkey switched on this trial and add it to the dataframe
behav = add_switch_info(behav)

```

```python
# print number of unique 'stay_value' s
behav['prev_switch'] = behav.groupby(['monkey', 'session'])['switch'].shift(1)
post_switch_trials = behav[behav['prev_switch'] == 1]
print(f"Number of unique 'stay_value' s on post-switch trials: {post_switch_trials['stay_value'].nunique()}")

for value in post_switch_trials['stay_value'].unique():
    count = post_switch_trials[post_switch_trials['stay_value'] == value].shape[0]
    print(value, count)
```
