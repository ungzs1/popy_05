```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Parameters

n = 50  # number of trials
alphas = [0.05, 0.1, 0.3, 0.5]  # different learning rates

# Past trials axis: 0 = most recent, 1 = one step back, ..., n-1 = oldest
past_trials = np.arange(0, n)

# Compute omega weights as function of distance into past
weights_past = {}
for alpha in alphas:
    omega = np.array([alpha * (1-alpha)**k for k in past_trials])
    weights_past[alpha] = omega

# Plot
plt.figure(figsize=(4, 3))
for alpha, omega in weights_past.items():
    plt.plot(past_trials, omega, label=f"$\\alpha={alpha}$")

plt.xlabel("Trials into the past (0 = most recent)")
plt.ylabel("Weight $\\omega$")
plt.title("Exponential weighting of past outcomes")
# despine
sns.despine()

plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)

# save figure
plt.tight_layout()
plt.savefig("figs/delta_rule_weights.png", dpi=300, bbox_inches='tight', transparent=True)

plt.show()
```
