# Summary

In almost every session we have a large proportion of units which do not spike all the way along the session, but either start or stop firing during in the middle of the session. This makes it imporrible to
do trial-by-trial analysis on the whole population, as this way we have missing datapoint. The solution is to remove units with truncated data, which leaves us with vary few units to analyze in many cases.

In this notebook I will show how much we remove for each session.

---
# Setup

```python
#%% Imports
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from popy.io_tools import * 
from popy.neural_data_tools import *
```

```python
##% Plotting functions
def plot_spike_trains(neural_data, monkey, session):
    neural_data['time'] = neural_data['time'] / 60
    plt.imshow(neural_data.spike_trains, aspect='auto', cmap='Grays', interpolation='none', 
               extent=[0, neural_data.time.values[-1], 0, len(neural_data.unit.values)],
               vmin=-1, vmax=0)
    
    ylabel = [f'{unit}' for unit in neural_data.unit.values]
    plt.yticks(np.arange(len(ylabel)), ylabel, fontsize=5)

    # add lables and legend: white horizontal bar is no recording, black is recording
    plt.plot([0, 0], [0, 0], color='black', label='Recording', linewidth=5)
    plt.plot([0, 0], [0, 0], color='white', label='No recording', linewidth=5)
    plt.legend(loc='upper right')

    plt.grid(alpha=.2)

    plt.xlabel('Time (mins)')
    plt.ylabel('Neurons')
    plt.title(f'{monkey}_{session}')
    plt.show()
```

---
# Main

```python
session_metadata = load_metadata()
session_metadata = session_metadata[session_metadata['block_len_valid']]

for _, session_info in session_metadata.iterrows():
    monkey, session = session_info['monkey'], session_info['session']

    neural_data = load_neural_data(monkey, session, hz=1000)

    # remove some units
    #neural_data = remove_low_fr_neurons(neural_data, 1, print_usr_msg=True)
    neural_data = remove_trunctuated_neurons(neural_data, mode='set_nan', delay_limit=100, print_usr_msg=True)

    # convert nans to -1, non-nans to 1
    neural_data['spike_trains'] = neural_data['spike_trains'] * 0
    neural_data = neural_data.fillna(-1)

    plot_spike_trains(neural_data, monkey, session)
    
```
